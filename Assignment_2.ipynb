```python
size = eval(input('enter the size of letter A'))
mid = size//2
for i in range (size):
    for j in range (size*2):
        if(
            j==size-1 or
            j==size+1 or
            (i==mid and size-i<j<size+i)
        ):
            print('*', end = ' ')
        else:
            print(' ',end=' ')
    print()
```
