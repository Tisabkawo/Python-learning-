```python
from math import floor
y = eval(input('Enter a year'))
c = y//100
m = (15+c-floor(c/4)-floor((8*c+13)/25))%30
n = (4+c-floor(c/4))%7
a = y%4
b = y%7
c = y%19
d = (19*c+m)%30
e = (2*a + 4*b + 6*d + n)%7
if d==29 and e==6:
    print('Easter is April 19')
elif d==28 and e==6 and m in [2,5,10,13,16,21,24,39]:
    print('Easter is April 18')
elif 22 + d +e<=31:
    print('Easter is March', 22+d+e)
else:
    print ('Easter is April', d+e-9)
```
